```python
import requests
```

```python
WEBHOOK_URL = "https://qyapi.weixin.qq.com/cgi-bin/webhook/send?key=d24cbb80-112d-4e18-a7ee-9561d50e3a67"
```

```python
def send_markdown(content: str):
    """发送 markdown 消息到企微群。"""
    payload = {
        "msgtype": "markdown",
        "markdown": {
            "content": content
        }
    }
    resp = requests.post(WEBHOOK_URL, json=payload, timeout=10)
    result = resp.json()
    if result.get("errcode") != 0:
        print(f"发送失败: {result}")
    return result
```

```python
send_markdown("## 我五月24回\n<@marrtinwang>")
```
